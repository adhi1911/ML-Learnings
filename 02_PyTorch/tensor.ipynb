
## The Golden Rule

* unsqueeze(): Adds a dimension of size 1. (Makes the tensor more complex/nested).
* squeeze(): Removes dimensions of size 1. (Flattens out unnecessary single-item dimensions).

------------------------------
## 1. unsqueeze(dim): Adding a Dimension
You use unsqueeze when a PyTorch model expects a specific number of dimensions, but your data is missing one. The dim parameter specifies where to insert the new dimension.
Imagine you have a 1D tensor representing 3 features:

x = torch.tensor([10, 20, 30])  # Shape: torch.Size([3]) (1D)

By unsqueezing at different positions, you change its structural meaning:

* x.unsqueeze(0) → Shape: torch.Size([1, 3]) (2D)
* What it means: It turns a list into a matrix with 1 row and 3 columns. (Commonly used to add a mock "Batch Size" of 1).
* x.unsqueeze(1) → Shape: torch.Size([3, 1]) (2D)
* What it means: It turns a list into a matrix with 3 rows and 1 column.

------------------------------
## 2. squeeze(): Removing Dimensions
If you have dimensions of size 1 that you don't need anymore, squeeze() drops them entirely.
Imagine you have a tensor with an annoying shape from an image batch extraction:

y = torch.randn(1, 3, 1, 32)  # Shape: torch.Size([1, 3, 1, 32])


* y.squeeze() (No arguments)
* Result: Shape becomes torch.Size([3, 32]).
   * What happened: PyTorch automatically looked for all dimensions that equal 1 and deleted them.
* y.squeeze(0) (Specific dimension)
* Result: Shape becomes torch.Size([3, 1, 32]).
   * What happened: It only removed the 0th dimension because it was size 1.
* y.squeeze(1) (Safe check)
* Result: Shape stays torch.Size([1, 3, 1, 32]).
   * What happened: Nothing! Dimension 1 has a size of 3, so PyTorch safely ignores it. squeeze() never crushes data greater than 1.
