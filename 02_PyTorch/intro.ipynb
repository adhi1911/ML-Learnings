# Automatic Differentiation Engine 
example of RNN 

```python
x = torch.randn(1,10)
prev_h = torch.randn(1,20)
W_h = torch.randn(20,20)
W_x = torch.randn(20,10)

i2h = torch.mm(W_x , x.t())
h2h = torch.mm(W_h, prev_h.t())
next_h = i2h + h2h 
next_h = next_h.tanh()

loss = next_h.sum()
```

Explanation of above model in short: 
x is the input tensor (1,10) and h is the hidden state tensor (1,20). The RNN takes these inputs and produces an output tensor (1,20) and an updated hidden state tensor (1,20). The output tensor represents the predictions made by the RNN based on the input data, while the hidden state tensor captures the internal memory of the RNN that is used to make future predictions.

dimensions : x is 1,10 because it represents a single sequence of length 10, while h is 1,20 because it represents the hidden state of the RNN, which has a size of 20. The output tensor is also 1,20 because it represents the predictions made by the RNN based on the input data and the hidden state.

# A simple model

```python
import torch.nn as nn 
import torch.nn.functional as F
```

```python
class LeNet(nn.Module):

    def __init__(self):
        super(LeNet, self).__init__()

        self.conv1 = nn.Conv2d(1,6,3)
        self.conv2 = nn.Conv2d(6,16,3)

        self.fe1 = nn.Linear(16*6*6, 120)
        self.fe2 = nn.Linear(120, 84)
        self.fe3 = nn.Linear(84, 10)

    def forward(self, x):
        x = F.max_pool2d(F.relu(self.conv1(x)), (2,2))
        x = F.max_pool2d(F.relu(self.conv2(x)), 2)
        x = x.view(-1 , self.num_flat_features(x))
        x = F.relu(self.fe1(x))
        x = F.relu(self.fe2(x))
        x = self.fe3(x)
        return x 

    def num_flat_features(self,x):
        size = x.size()[1:]
        num_features = 1
        for s in size:
            num_features *= s
        return num_features
```

```python
net = LeNet()
print(net)
```

```python
input = torch.randn(1,1,32,32)
print(net(input))
```
