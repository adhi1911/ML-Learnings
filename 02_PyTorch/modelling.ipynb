````python
import torch 

class Net(torch.nn.Module):

    def __init__(self):
        super(Net,self).__init__()

        self.linear1 = torch.nn.Linear(100,200)
        self.activation = torch.nn.ReLU()
        self.linear2 = torch.nn.Linear(200,10)
        self.softmax = torch.nn.Softmax(dim=1)

    def forward(self, x):
        x = self.linear1(x)
        x = self.activation(x)  
        x = self.linear2(x)
        x = self.softmax(x)

        return x 
````

````python
net = Net()

print(net)
# print(net.linear1.bias, net.linear1.weight)
print(net.linear1.children)
print(net.linear2)
````

````python
for param in net.parameters():
    print(param)
    print(param.shape)
````


* nn.ReLU is an Object/Module. It acts like a structural building block.
* F.relu (from torch.nn.functional) is a Pure Function. It acts like a raw mathematical operator.

------------------------------
## 1. nn.ReLU (The Object/Layer Approach)
When you use nn.ReLU, you are creating a Python object (an instance of a class). Because it is an object, it can be dropped into structural containers like nn.Sequential.
How you use it:

```python
import torch.nn as nn
class MyModel(nn.Module):
    def __init__(self):
        super().__init__()
        # You define it as a structural layer in __init__
        self.layer1 = nn.Linear(10, 20)
        self.relu = nn.ReLU() 

    def forward(self, x):
        # You call it like a layer in forward
        x = self.layer1(x)
        x = self.relu(x)
        return x
```

* Best for: Clean, structured code or when building models using nn.Sequential.

------------------------------
## 2. F.relu (The Function Approach)
When you use F.relu, you are calling a raw function directly. It doesn't hold any state, it doesn't need to be initialized in __init__, and you just import it and use it right inside the forward pass.
How you use it:
```python
import torch.nn as nnimport torch.nn.functional as F
class MyModel(nn.Module):
    def __init__(self):
        super().__init__()
        # No ReLU defined here! Only layers with learnable weights.
        self.layer1 = nn.Linear(10, 20)

    def forward(self, x):
        # You apply it directly like a math function
        x = self.layer1(x)
        x = F.relu(x) 
        return x
```

* Best for: Quick operations, avoiding clutter in your __init__, and activations that don't have any weights to learn.

------------------------------
## The Golden Rule of Thumb: Do they have weights?
Activations like ReLU, Softmax, Sigmoid, and Tanh do not have any learnable weights (parameters). They just take a number and change it based on a formula. Because they have no weights, using F.relu or nn.ReLU makes absolutely zero difference to your model's performance or training. It is purely a matter of personal coding style.
However, for layers that do have weights—like nn.Linear (Linear/Dense layers) or nn.Conv2d (Convolutional layers)—you must use the nn object version in __init__ so PyTorch knows to track and update those weights during backpropagation.
