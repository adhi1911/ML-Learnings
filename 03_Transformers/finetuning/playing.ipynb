# Transformers
Transformers are a type of deep learning model that have revolutionized the field of natural language processing (NLP) and have also been applied to other domains such as computer vision and speech recognition. They were introduced in the paper "Attention is All You Need" by Vaswani et al. in 2017.

## Transformer Architecture

The transformer architecture is based on the concept of self-attention, which allows the model to weigh the importance of different words in a sentence when making predictions. The main components of a transformer model include:

1. **Encoder**: The encoder processes the input sequence and generates a representation of it. It consists of multiple layers, each containing a self-attention mechanism and a feed-forward neural network.
2. **Decoder**: The decoder generates the output sequence based on the encoder's representation. It also consists of multiple layers, each containing a self-attention mechanism, a feed-forward neural network, and an additional attention mechanism that attends to the encoder's output.
3. **Attention Mechanism**: The attention mechanism allows the model to focus on different parts of the input sequence when generating the output. It computes a weighted sum of the input representations, where the weights are determined by the relevance of each input token to the current output token being generated.

4. **Positional Encoding**: Since transformers do not have a built-in notion of word order, positional encoding is added to the input embeddings to provide information about the position of each token in the sequence.
5. **Multi-Head Attention**: This allows the model to attend to different parts of the input sequence simultaneously, capturing various aspects of the input data.



### Transformer Data Flow Diagram

**Encoder:**
1. **Input Text**  
    ↓  
2. **Tokenization**  
    ↓  
3. **Token Encoding**  
    ↓  
4. **Token Embedding**  
    ↓  
5. **Encoder Stack (Multiple Layers with Self-Attention & Feed-Forward)**  
    ↓  
6. **Hidden States (Contextual Representations)**

---

**Decoder:**
1. **Previous Output Tokens**  
    ↓  
2. **Token Encoding**  
    ↓  
3. **Token Embedding**  
    ↓  
4. **Decoder Stack (Multiple Layers with Self-Attention, Encoder-Decoder Attention & Feed-Forward)**  
    ↓  
5. **Output Tokens**

### Somethings to know about:
- Tokenized text: The input text is tokenized into smaller units (tokens) that can be processed by the model. This can be done using various tokenization techniques, such as word-level, subword-level, or character-level tokenization.
- Token Encodings: Each token is assigned a unique integer ID based on a predefined vocabulary. This process is known as token encoding, and it allows the model to work with numerical data instead of raw text.
- Token Embeddings: Each token is converted into a dense vector representation (embedding) that captures its semantic meaning. These embeddings are learned during the training process and can be fine-tuned for specific tasks.
- Encode Stack: The encoder consists of multiple layers that process the input embeddings and generate a contextual representation of the input sequence. Each layer includes a self-attention mechanism and a feed-forward neural network.

We can check it using tokenizer's max_length attribute. If we try to encode a sentence that is longer than the max_length, it will be truncated to fit the max_length. ## Tokenization of our data

```python
ds.reset_format()

ds
```

```python
def tokenize(batch): 
    
    tokenized_text = tokenizer(batch['text'], padding=True, truncation= True)

    return tokenized_text
```

```python
ds_encoded = ds.map(tokenize, batched=True, batch_size=None)
```

## Model Building

```python
from transformers import AutoModel 
import torch 
```

tokenizer with return_tensors parameter set to "pt" will return the tokenized output as a PyTorch tensor. This is useful when we want to feed the tokenized input directly into a PyTorch model for training or inference. The output will be a dictionary containing the input_ids, attention_mask, and other relevant information in the form of PyTorch tensors.

without this param , the output will be a dictionary containing the input_ids, attention_mask, and other relevant information in the form of lists. This is useful when we want to inspect the tokenized output or use it for other purposes that do not require PyTorch tensors. This is similar to the output above where we have the input_ids and attention_mask as lists instead of tensors.

```python
tokenizer("Hello macha", return_tensors="pt")
```
